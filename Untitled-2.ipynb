```python
# Basic imports
import os
import numpy as np
import pandas as pd
import re
from pathlib import Path
from tqdm.auto import tqdm

# ML libraries
import lightgbm as lgb
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# Image processing
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.preprocessing import image
from PIL import Image

# Device info (optional, for torch if used later)
import torch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# Folders for images & embeddings
os.makedirs("images_75k", exist_ok=True)
os.makedirs("embeddings", exist_ok=True)
```

```python
# Log-transform target
df['log_price'] = np.log1p(df['price'])

# Extract quantity from catalog content
def extract_quantity(text):
    match = re.search(r'(\d+(\.\d+)?)\s*(oz|g|ml|lb|pack|ct)', str(text), re.I)
    return float(match.group(1)) if match else np.nan

df['quantity'] = df['catalog_content'].apply(extract_quantity)
df['quantity'].fillna(df['quantity'].median(), inplace=True)
```

```python
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2))
X_text = tfidf.fit_transform(df['catalog_content'])
y = df['log_price'].values
print("TF-IDF shape:", X_text.shape)
```

```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import numpy as np
from tqdm.auto import tqdm

# GPU device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Load ResNet50 without classifier (matches TF include_top=False)
resnet = models.resnet50(weights="IMAGENET1K_V2")
resnet = torch.nn.Sequential(*list(resnet.children())[:-1])  # remove top layer
resnet = resnet.to(device)
resnet.eval()

# Image preprocessing transform
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])
```

```python
BATCH_SIZE = 32  # adjust if GPU memory allows
features = []

for i in tqdm(range(0, len(df), BATCH_SIZE)):
    batch_paths = [f"images_75k/{sid}.jpg" for sid in df['sample_id'].iloc[i:i+BATCH_SIZE]]
    batch_imgs = []
    
    for p in batch_paths:
        try:
            img = Image.open(p).convert("RGB")
            batch_imgs.append(transform(img))
        except:
            batch_imgs.append(torch.zeros(3,224,224))  # fallback zero image

    x = torch.stack(batch_imgs).to(device)
    
    with torch.no_grad():
        batch_features = resnet(x)              # shape: (batch_size, 2048,1,1)
        batch_features = batch_features.view(batch_features.size(0), -1)  # flatten 2048-dim
        batch_features = batch_features.cpu().numpy()
    
    features.append(batch_features)

# Combine all batches
X_image = np.vstack(features)
print("Image features shape:", X_image.shape)

# Optional: save for later to avoid recomputation
np.save("X_image_40k.npy", X_image)
```

```python
from sklearn.model_selection import train_test_split
import lightgbm as lgb
from scipy.sparse import hstack

# TF-IDF + numerical features (example: quantity column if exists)
if 'quantity' in df.columns:
    X_num = df[['quantity']].values
    X_train_final = hstack([X_text, X_num])
else:
    X_train_final = X_text

# Split for validation
X_tr, X_val, y_tr, y_val = train_test_split(X_train_final, df['log_price'].values, test_size=0.2, random_state=42)
```

```python
text_model = lgb.LGBMRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=8,
    random_state=42
)
text_model.fit(X_tr, y_tr)
```

```python
image_model = lgb.LGBMRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=8,
    random_state=42
)
image_model.fit(X_image, df['log_price'].values)
```

```python
def smape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return 100 * np.mean(np.abs(y_pred - y_true) / ((np.abs(y_true) + np.abs(y_pred)) / 2))

# Predict
y_pred_text = text_model.predict(X_train_final)
y_pred_image = image_model.predict(X_image)

# Weighted ensemble
y_pred_ensemble = 0.6 * y_pred_text + 0.4 * y_pred_image

# Evaluate
smape_score = smape(df['log_price'].values, y_pred_ensemble)
print("Weighted Ensemble SMAPE:", smape_score)
```

```python
import pandas as pd
import numpy as np

# Number of samples to display
n_samples = 100

# Random indices
sample_indices = np.random.choice(len(df), size=n_samples, replace=False)

# Actual prices (reverse log1p to original scale)
y_actual = np.expm1(df['log_price'].values)

# Predicted prices (reverse log1p if needed)
y_pred = np.expm1(y_pred_ensemble)

# Create comparison DataFrame
comparison_df = pd.DataFrame({
    'Sample_ID': df['sample_id'].iloc[sample_indices].values,
    'Actual Price': y_actual[sample_indices],
    'Predicted Price': y_pred[sample_indices],
    'Difference': y_actual[sample_indices] - y_pred[sample_indices]
})

# Optional: sort by difference
comparison_df = comparison_df.sort_values(by='Difference', key=abs, ascending=False)

# Display
pd.set_option('display.float_format', lambda x: '%.2f' % x)
print(comparison_df)
```

```python
def download_image(url, path):
    try:
        r = requests.get(url, timeout=10)
        if r.status_code == 200:
            with open(path, 'wb') as f:
                f.write(r.content)
            return True
    except:
        pass
    return False

for idx, row in tqdm(df_test.iterrows(), total=len(df_test), desc="Downloading images"):
    img_path = Path("test_images")/f"{int(row['sample_id'])}.jpg"
    if not img_path.exists():
        download_image(row['image_link'], img_path)
```
